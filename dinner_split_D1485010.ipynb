作業說明：
這次作業使用 Python 製作聚餐費用試算器。
先將文字資料整理並轉換成數字，再計算服務費、
折價後的總金額和每人分攤費用，最後判斷是否超過預算。

測試一：4 人，折價券 100 元
測試二：5 人，折價券 100 元
測試三：4 人，折價券 220 元

```python

# 測試一：原始資料
event_name = "  WEEKEND DINNER  "
meal_amount = " NT$1,200 "
people = "4"

service_rate = 0.10
discount = 100
budget_per_person = 300

# 整理文字資料
clean_name = event_name.strip().lower()
clean_amount = meal_amount.replace("NT$", "").replace(",", "").strip()

# 將文字轉成數字
meal_price = float(clean_amount)
people_count = int(people)

# 計算聚餐費用
service_fee = meal_price * service_rate
total_amount = meal_price + service_fee - discount
cost_per_person = total_amount / people_count

# 判斷是否超過預算
over_budget = cost_per_person > budget_per_person

# 輸出結果
print("聚餐名稱：", clean_name)
print(f"原始餐費：{meal_price:.2f} 元")
print(f"服務費：{service_fee:.2f} 元")
print(f"折價券：{discount:.2f} 元")
print(f"應付總額：{total_amount:.2f} 元")
print(f"分攤人數：{people_count} 人")
print(f"每人費用：{cost_per_person:.2f} 元")
print(f"每人預算：{budget_per_person:.2f} 元")
print("是否超過預算：", over_budget)
```
